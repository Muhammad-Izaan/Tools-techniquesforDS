```python
def fun(num1, num2):
    return num1*num2
```

```python
fun(2, 3)
```

```python
def check_vowels(text, vowels='aeiou'):
    vowel_count=0

    for char in text:
        if char in vowels:
            print(f"'{char}'")
            vowel_count += 1
        else:
            print(char)
    print(f"There are a total of {vowel_count} vowels in the '{text}'")

check_vowels("hello dear!")
```

Lambda Function

```python
num_sqr=lambda x:x*x
```

```python
num_sqr(2)
```

```python
even_odd= lambda num: "even" if num %2==0 else "odd"
```

```python
even_odd(21)
```

```python
cal = lambda a, b=5, c=6: (a*b)-c
cal(10)
```
