```python
# Filters
# filters finds the condition true false
# maps prints the body of whatever condition is true
# startswith is python method called for string only
# isinstance() → Checks datatype.
```

```python
mix_list=["ali", True, "sara", "ahmed", False, "bilal", 58,963,52.34, "", bool]

list(filter(lambda str_val
           : isinstance(str_val, bool),
           mix_list))
```

```python
list(filter(lambda str_val
           : isinstance(str_val, str) and str_val != "",
           mix_list))
```

```python
# functools is a library

from functools import reduce

numbers = [10, 20, 30, 40]

result = reduce(lambda a, b: a + b, numbers)

print(result)
```

```python
numbers = [53, 640, 204, 95,963,124,857]

result = reduce(lambda a, b: a if a>b else b, numbers)

print(result)
```

```python
# reduce is a function which we are importing from functools library

names=['ali', 'sara', 'ahmed', 'bilal']

longest_name=reduce(lambda a, b: a if len(a)> len(b) else b, names)
print(longest_name)
```

```python
student_marks=[78,54,89,90,61,98,36]
pass_marks=[]
fail_marks=[]

for i in student_marks:
    if i <= 50:
        fail_marks.append(i)
    else:
        pass_marks.append(i)
print(f"Total Failed Student: len{fail_marks}, {fail_marks}")
print(f"Total Pass Student: len{pass_marks}, {pass_marks}")
```

```python
items_price= [1200,450,1200,3000,450,150,3000,603,150,890]
unique=[]
duplicates=[]

grand_total=0

for i in items_price:
    if i not in unique:
        unique.append(i)
    else:
        duplicates.append(i)
for j in unique:
    if j >= 1000:
        grand_total +=j

print("Unique Total: ", unique)
print("Duplicate Total: ", duplicates)
print("Grand Total: ", grand_total)
```
