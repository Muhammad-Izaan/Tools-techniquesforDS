```python
runningSum=0
counter=0

while counter !=3:
    num=int(input("Enter a number or 0 to exit"))

    if num != 0:
        runningSum+=num
        counter+=1
    else:
        print("the loop has been broken")
        break
    print(runningSum)
```

```python
balance=5000

while True:
    amount = int(input("enter withdrawal amount"))
    if amount == 0:
        print("nothing deducted")
        break

     elif amount > balance:
        print("insufficient balance to proceed")
    else:
        balance=balance-amount
        print("your current balance is=", balance)

    if balance == 0:
        break

print("funds finish")
```
